# Notebook 04 — Analysis and Figures

Reads `results/el_gzip_results.csv`, produces:
- `results/figures/fig1_mvp_el_gzip.png` — hero image
- `results/minimum_viable_result_check.txt`
- Per-level summary table with 95% bootstrap CI

**Rule:** EL_gzip is never described as the CTW-normalized Solomonoff Gap.

```python
import sys
from pathlib import Path
import pandas as pd

repo_root = Path('.').resolve().parent
sys.path.insert(0, str(repo_root / 'src'))

results_csv = repo_root / 'results' / 'el_gzip_results.csv'
figures_dir = repo_root / 'results' / 'figures'
figures_dir.mkdir(exist_ok=True)

df = pd.read_csv(results_csv)
print('Loaded:', len(df), 'rows')
print('Models:', df['model'].unique().tolist())
print('Levels:', sorted(df['complexity_level'].unique().tolist()))
```

```python
from solomonoff_bench.analysis.stats import level_summary, minimum_viable_result_check

summary = level_summary(df)
print(summary.to_string(index=False))
```

```python
# Gzip diagnostics — negative deltas
neg = df[df['h_gzip_is_negative'] == True]
print('Negative gzip deltas:', len(neg), '/', len(df),
      '=', round(100 * len(neg) / len(df), 2), '%')
if len(neg) > 0:
    print('  Note: logged and reported. Not clamped. Justifies CTW in Week 2.')
```

```python
# Invalid-token mass diagnostics
for model in df['model'].unique():
    m = df[df['model'] == model]
    print('Model:', model)
    print('  Mean invalid-token mass:', round(m['mean_invalid_mass'].mean(), 4))
    print('  Mean valid binary mass: ', round(m['mean_valid_binary_mass'].mean(), 4))
```

```python
from solomonoff_bench.analysis.plots import plot_figure1_el_gzip

fig_path = figures_dir / 'fig1_mvp_el_gzip.png'
plot_figure1_el_gzip(results_csv, fig_path)

from IPython.display import Image
Image(str(fig_path))
```

```python
mvr_path = repo_root / 'results' / 'minimum_viable_result_check.txt'
text = minimum_viable_result_check(df, mvr_path)
print(text)
```
