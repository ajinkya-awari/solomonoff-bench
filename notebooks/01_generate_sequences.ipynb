# Notebook 01 — Generate 300 MVP Sequences

Runs on CPU (no GPU needed). Produces `data/sequences_mvp.json`.

**Complexity levels:** 4 levels × 75 unique sequences × 200 symbols each = 300 total.  
**TM convention:** output-transducer — written symbol emitted at every transition.

```python
import sys
from pathlib import Path
repo_root = Path('.').resolve().parent
sys.path.insert(0, str(repo_root / 'src'))

from solomonoff_bench.sequences.generate_sequences import generate_mvp_dataset, validate_dataset

output_path = repo_root / 'data' / 'sequences_mvp.json'
records = generate_mvp_dataset(output_path=output_path, verbose=True)
validate_dataset(records)
print('Done.')
```

```python
import json
with open(output_path) as f:
    ds = json.load(f)

print('Total sequences:', ds['total_sequences'])
print('Output convention:', ds['output_convention'])
print()
for level, info in ds['complexity_levels'].items():
    recs = [r for r in ds['records'] if r['complexity_level'] == int(level)]
    print(f"  Level {level}: n_states={info['n_states']}, program_bits={info['program_bits']}, sequences={len(recs)}")

sample = ds['records'][0]
print()
print('Sample record keys:', list(sample.keys()))
print('Sample sequence (first 40 chars):', sample['sequence'][:40], '...')
```
