For each model, builds the binary token map and prints the full validation table.

```python
import sys
from pathlib import Path
repo_root = Path('/kaggle/working/solomonoff-bench')  # adjust if local
sys.path.insert(0, str(repo_root / 'src'))

from transformers import AutoTokenizer
from solomonoff_bench.models.tokenizer_validation import build_binary_token_map

MODELS = [
    'microsoft/Phi-3-mini-4k-instruct',
    # 'meta-llama/Llama-3.2-3B-Instruct',  # uncomment if HF_TOKEN set
    # 'TinyLlama/TinyLlama-1.1B-Chat-v1.0',
]

for model_name in MODELS:
    print('=' * 60)
    print('Model:', model_name)
    tok = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
    tm = build_binary_token_map(tok, model_name)
    s = tm.summary()
    print('  Valid zero token IDs:', s['zero_token_ids'])
    print('  Valid one  token IDs:', s['one_token_ids'])
    print('  Zero forms:', s['zero_forms'])
    print('  One  forms:', s['one_forms'])
    print('  Total vocab size:', s['total_vocab_size'])
    print('  is_valid:', tm.is_valid())
    if not tm.is_valid():
        print('  WARNING: no valid binary tokens found!')
    print()
```
